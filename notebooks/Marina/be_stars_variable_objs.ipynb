```python
import pandas as pd
import numpy as np
import matplotlib.colors as colors

from matplotlib import cm
from matplotlib.patches import Polygon
from matplotlib import pyplot as plt

from astropy.io import fits
from astropy import units as u
from astropy.coordinates import SkyCoord
from astropy.coordinates import BaseEclipticFrame

import copy
import cartopy.crs as ccrs


def plot_splus_field(ra, dec, galactic = False, ax = plt, **kargs):
    
    if ra > 180:
        ra = ra - 360
    ra = -ra
    
    # Calculate corners
    dec0 = dec - 0.7
    dec1 = dec - 0.7
    dec2 = dec + 0.7
    dec3 = dec + 0.7

    ra0 = ra - 0.7/np.cos(dec0*np.pi/180)
    ra1 = ra + 0.7/np.cos(dec1*np.pi/180)
    ra2 = ra + 0.7/np.cos(dec2*np.pi/180)
    ra3 = ra - 0.7/np.cos(dec3*np.pi/180)
    
    ra = np.array([ra0, ra1, ra2, ra3, ra0])
    dec = np.array([dec0, dec1, dec2, dec3, dec0])
    
    ra[ra > 180] = ra[ra > 180] - 360   
    
    dec[dec > 90] = 90
    dec[dec < -90] = -90
    
    ax.fill(ra, dec, **kargs)

# Get the galactic plane

gal_ll = np.arange(-180, 180, 0.1)
gal_bb = np.full(len(gal_ll), 0)

c = SkyCoord(frame="galactic", l=gal_ll*u.degree, b=gal_bb*u.degree)

gal_ra = c.icrs.ra.degree
gal_de = c.icrs.dec.degree

gal_ra[gal_ra > 180] = gal_ra[gal_ra > 180] - 360

# Get ecliptic plane

ecl_lon = np.arange(-180, 180, 0.1)
ecl_lat = np.full(len(gal_ll), 0)

ecl_c = SkyCoord(frame='geocentrictrueecliptic', lon=ecl_lon*u.degree, lat=ecl_lat*u.degree)

ecl_ra = ecl_c.icrs.ra.degree
ecl_de = ecl_c.icrs.dec.degree

ecl_ra[ecl_ra > 180] = ecl_ra[ecl_ra > 180] - 360

# Get equator
eq_ra = np.arange(-180, 180, 0.1)
eq_de = np.full(len(eq_ra), 0)

# Get galactic poles

sgp_ll = 0
sgp_bb = -90

c = SkyCoord(frame="galactic", l=sgp_ll*u.degree, b=sgp_bb*u.degree)

sgp_ra = c.icrs.ra.degree
sgp_de = c.icrs.dec.degree

ngp_ll = 0
ngp_bb = 90

c = SkyCoord(frame="galactic", l=ngp_ll*u.degree, b=ngp_bb*u.degree)

ngp_ra = c.icrs.ra.degree
ngp_de = c.icrs.dec.degree

gctr_ll = 0
gctr_bb = 0

c = SkyCoord(frame="galactic", l=gctr_ll*u.degree, b=gctr_bb*u.degree)

gctr_ra = c.icrs.ra.degree
gctr_de = c.icrs.dec.degree

lmc_ra = 15*(5+23/60.)
lmc_de = -1*(69+45/60.)

smc_ra = 15*(0+52/60.)
smc_de = -1*(72+48/60.)


def plot_grid(ax = plt, **kwargs):
    # Plot lines
    for ra in np.arange(-180, 180, 30):
        dera = np.arange(-90, 90, 0.1)
        rara = np.full(len(dera), ra)
        ax.plot(-rara, dera, transform=ccrs.Geodetic(), color = "#000000", alpha = 0.5, linewidth = 0.5, zorder = 3)

    # Plot lines
    for de in np.arange(-90, 90, 30):
        rade = np.arange(-180, 180, 0.1)
        dede = np.full(len(rade), de)
        ax.plot(-rade, dede, transform=ccrs.Geodetic(), color = "#000000", alpha = 0.5, linewidth = 0.5, zorder = 3)

    # plot labels
    ra_tick_label = [-150, -120, -90, -60, -30, 0, 30, 60, 90, 120, 150]
    ra_text_label = ['14h', '16h', '18h', '20h', '22h', '0h', '2h', '4h', '6h', '8h', '10h']

    for ra_tick, ra_text in zip(ra_tick_label, ra_text_label):
        ax.text(-ra_tick, 3, ra_text, fontsize=16, transform=ccrs.Geodetic(), **kwargs)

    dec_tick_label = [-60, -30, 0, 30, 60]
    dec_text_label = ['-60°   ', '-30°  ', '0° ', '30° ', '60° ']

    for de_tick, de_text in zip(dec_tick_label, dec_text_label):
        ax.text(-180, de_tick, de_text, ha='right', va='center', fontsize=16, transform=ccrs.Geodetic(), **kwargs)

plt.figure(figsize=(12,6))
#projection = ccrs.InterruptedGoodeHomolosine()
#projection = ccrs.Miller()
#projection = ccrs.Mercator()
projection = ccrs.Mollweide()
#projection = ccrs.Robinson()
#projection = ccrs.EckertIII()
#projection = ccrs.EqualEarth()

ax = plt.axes(projection=projection)

# Plot galactic plane
ax.plot(-gal_ra, gal_de, transform=ccrs.Geodetic(), color = "#00ffff", lw = 2, zorder = 10, alpha = 0.2)

ax.scatter(-sgp_ra, sgp_de, transform=ccrs.Geodetic(), color = "#ff00ff", marker="x", s = 100, alpha = 0.7, zorder = 21)
ax.scatter(-ngp_ra, ngp_de, transform=ccrs.Geodetic(), color = "#ff00ff", marker="x", s = 100, alpha = 0.7, zorder = 21)
ax.scatter(-gctr_ra, gctr_de, transform=ccrs.Geodetic(), color = "#00ffff", marker="x", s = 100, alpha = 0.9, zorder = 21)

# Plot ecliptic plane
ax.plot(-ecl_ra, ecl_de, transform=ccrs.Geodetic(), color = "#aaaa00", lw = 2, zorder = 11, alpha = 0.2)

# Plot equator plane
ax.plot(-eq_ra, eq_de, transform=ccrs.Geodetic(), color = "#0066aa", lw = 2, zorder = 12, alpha = 0.2)

# Extent em (lon, lat) geodésicos; aqui uso algo amplo ao redor LMC/SMC
ax.set_extent([-80, 30, -85, -40], crs=ccrs.Geodetic())

def plot_points_on_footprint(df, ra_col="ra", dec_col="dec",
                             ax=None, color="red", size=20, zorder=100,
                             transform=ccrs.Geodetic(), **kwargs):
    """
    Plota pontos de uma tabela (df) contendo RA e DEC
    sobre o mapa já existente.

    df       : DataFrame pandas com colunas de RA/DEC
    ra_col   : nome da coluna de RA (em graus)
    dec_col  : nome da coluna de DEC (em graus)
    ax       : eixo cartopy já existente (se None, usa plt.gca())
    color    : cor dos marcadores
    size     : tamanho dos pontos
    zorder   : camada do plot
    transform: transformação (use Geodetic com projeção Mollweide)
    kwargs   : parâmetros extras para ax.scatter()
    """

    if ax is None:
        ax = plt.gca()

    # RA precisa ser invertido para o estilo "sky map"
    ra = df[ra_col].values.copy()
    dec = df[dec_col].values.copy()

    # Converte RA > 180 → RA - 360
    ra = np.where(ra > 180, ra - 360, ra)

    # Inverte eixo do RA para o mapa celeste
    ra = -ra

    ax.scatter(ra, dec,
               s=size,
               color=color,
               transform=transform,
               zorder=zorder,
               **kwargs)
    
# Plot SPLUS footprint
for i in range(len(all_splus)):
    plot_splus_field(all_splus.loc[i,"ra"],all_splus.loc[i,"dec"],transform=ccrs.Geodetic(),
                     color = "#666666", alpha = 0.15, zorder = 50)

plot_points_on_footprint(
    df,       # seu DataFrame
    ra_col="ra",      # nome da coluna de RA
    dec_col="dec",    # nome da coluna de DEC
    ax=ax,            # o mesmo eixo Mollweide do footprint
    color="blue",
    size=1,
    zorder=200,
    alpha=0.8
)




plt.tight_layout()
```
