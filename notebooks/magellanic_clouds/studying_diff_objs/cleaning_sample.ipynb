```python
def unique_by_sep_with_score(ra_deg, dec_deg, score, min_sep_arcsec=2.0):
    ra_deg = np.asarray(ra_deg, float)
    dec_deg = np.asarray(dec_deg, float)
    score = np.asarray(score, float)

    order = np.argsort(score)[::-1]  # maior score primeiro
    inv = np.empty_like(order)
    inv[order] = np.arange(len(order))

    c = SkyCoord(ra=ra_deg*u.deg, dec=dec_deg*u.deg, frame="icrs")
    seplim = (min_sep_arcsec * u.arcsec)

    keep = np.ones(len(c), dtype=bool)

    # processa em ordem de score
    for k in order:
        if not keep[k]:
            continue
        # marca todos os vizinhos próximos (inclui ele mesmo) como removidos
        idx_close = c[k].search_around_sky(c, seplim)[0]
        idx_close = np.asarray(idx_close, int)
        idx_close = idx_close[idx_close != k]
        keep[idx_close] = False
        keep[k] = True

    return keep
```
